### Executive Summary & Business Impact
Money laundering operations across international borders introduce critical financial compliance risks. This project establishes an end-to-end Machine Learning and Graph Analysis pipeline designed to detect complex laundering schemes—specifically targeting **Fan-Out dispersal** and **Circular Flow layering** topologies.

By deploying an out-of-time temporal split, feature engineering targeting transactional burstiness, and class-weighted gradient boosting, our framework optimizes the **Area Under the Precision-Recall Curve (AUPRC)**. This minimizes false-positive alert fatigue for compliance officers while maintaining high recall on high-value illicit transfers.

### Stage 1: Domain Research, EDA & Problem Framing 
* **Laundering Mechanics:** Mapped transactions across Placement, Layering, and Integration stages.
* **Validation Split:** Implemented temporal split on transaction timestamps to prevent lookahead leakage.
* **Primary Metrics:** Selected AUPRC and minority-class F1-score over accuracy due to extreme class imbalance.

### Stage 2: Feature Engineering 
* **Account Dynamics:** Extracted in/out transaction ratios, unique counterparty counts, and cross-border currency indicators.
* **Pattern Signals:** Calculated rolling-window transaction burstiness and flagged round-number threshold values.

### Stage 3: Baseline Model Training 
* **Model Selection:** Trained class-weighted XGBoost to prioritize low-frequency positive illicit class instances.
* **Tuning Strategy:** Optimized tree depth, learning rate, and `scale_pos_weight` parameters for stable convergence.

### Stage 4: Evaluation, Cost Trade-off & SHAP Explainability 
* **Explainability:** Applied SHAP values to confirm feature importances (burstiness, counterparty count, amount dynamics).
* **Cost Analysis:** Balanced operational investigation costs (false alarms) against financial non-compliance penalties.
