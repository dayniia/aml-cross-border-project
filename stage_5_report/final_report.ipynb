### Executive Summary
This report consolidates the end-to-end Machine Learning pipeline for detecting illicit cross-border financial transactions. By combining temporal feature engineering, class-weighted gradient boosting, and graph topology visualization, our system flags high-risk laundering operations while mitigating compliance alert fatigue.

### Section 1: Domain Framing & Temporal EDA (Person 1 Hand-Off)
* **Laundering Stages:** Placement, Layering, and Integration.
* **Validation Strategy:** Out-of-time temporal train/test split to prevent future data leakage.
* **Headline Metric:** Area Under the Precision-Recall Curve (AUPRC) and Recall at fixed False Positive budgets, addressing heavy class imbalance.

### Section 2: Feature Engineering (Person 2 Hand-Off)
* **Transactional Features:** In/Out transaction ratios, cross-currency flags, and unique counterparty counts.
* **Behavioral Patterns:** Transaction burstiness within rolling time windows and round-number threshold indicators.

### Section 3: Model Performance & Explainability (Persons 3 & 4 Hand-Off)
* **Baseline Model:** Class-weighted XGBoost.
* **Explainability:** SHAP global feature importances confirm transaction frequency, burstiness, and counterparty volume as primary drivers for illicit predictions.
